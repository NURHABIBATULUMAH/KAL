# Langkah-Langkah SVD

### Matriks Awal

$$
A = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$

### Mencari Transpose A

Transpose = baris jadi kolom:

$$
A^T = \begin{bmatrix}
3 & 1 \\
1 & 3
\end{bmatrix}
$$

Karena $A$ simetris, maka $A^T = A$.

### Hitung $A^T A$


$$A^T A = A A =
\begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}
\begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}
=\begin{bmatrix}
10 & 6 \\
6 & 10
\end{bmatrix}$$

### Cari Nilai Eigen $A^T A$

$$
\det(A^T A - \lambda I) = 0
$$

$$
\det \begin{bmatrix}
10 - \lambda & 6 \\
6 & 10 - \lambda
\end{bmatrix} = 0
$$

$$
(10 - \lambda)^2 - 36 = 0
$$

$$
(10 - \lambda)^2 = 36
$$

$$
10 - \lambda = \pm 6
$$

$$
\lambda = 10 \mp 6
$$

$$
\Rightarrow \lambda_1 = 16, \quad \lambda_2 = 4
$$

### Singular Value

$$
\sigma_1 = \sqrt{16} = 4, \quad \sigma_2 = \sqrt{4} = 2
$$

### Cari Vektor Eigen untuk $V$

Untuk $\lambda = 16$:

$$
\begin{bmatrix} -6 & 6 \\ 6 & -6 \end{bmatrix}
\Rightarrow v_1 = v_2
$$

Normalisasi:

$$
v_1 = \frac{1}{\sqrt{2}}, \quad v_2 = \frac{1}{\sqrt{2}}
$$

Untuk $\lambda = 4$:

$$
\begin{bmatrix} 6 & 6 \\ 6 & 6 \end{bmatrix}
\Rightarrow v_1 = -v_2
$$

Normalisasi:

$$
v_1 = \frac{1}{\sqrt{2}}, \quad v_2 = -\frac{1}{\sqrt{2}}
$$

### Matriks V

$$
V = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

### Cari U

$$
U = A V \Sigma^{-1}
$$

$$
A V =
\begin{bmatrix}
2 \sqrt{2} & \sqrt{2} \\
2 \sqrt{2} & -\sqrt{2}
\end{bmatrix}
$$

Normalisasi:

$$
U = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$

### Hasil Akhir SVD

$$
A = U \Sigma V^T
$$

dengan:

$$
U = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}, \quad
\Sigma = \begin{bmatrix}
4 & 0 \\
0 & 2
\end{bmatrix}, \quad
V = \begin{bmatrix}
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} \\
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}}
\end{bmatrix}
$$
