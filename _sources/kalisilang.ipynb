## Perkalian Silang

## Definisi Hasil Kali Silang



$\vec{u} = \begin{bmatrix} u_1 \\ u_2 \\ u_3 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix}$

adalah vektor-vektor di $\mathbb{R}^3.$

Hasil kali silang dari $\vec{u}$ dan $\vec{v}$ dinotasikan sebagai $\vec{u} \times \vec{v}$ adalah:


$\vec{u} \times \vec{v} = 
\begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
u_3 v_1 - u_1 v_3 \\
u_1 v_2 - u_2 v_1
\end{bmatrix}$





## Contoh Penghitungan Hasil Kali Silang

Misalkan

$\vec{u} = \begin{bmatrix} 2 \\ -1 \\ 4 \end{bmatrix} \quad \text{dan} \quad
\vec{v} = \begin{bmatrix} 3 \\ 2 \\ 5 \end{bmatrix}$


Temukan $\vec{u} \times \vec{v}$, dan verifikasi bahwa hasil kali silang ini tegak lurus terhadap baik $\vec{u} \) maupun \( \vec{v}.$

## Solusi

Menggunakan definisi:


$\vec{u} \times \vec{v} =
\begin{bmatrix}
(-1)(5) - (4)(2) \\
-((2)(5) - (4)(3)) \\
(2)(2) - (-1)(3)
\end{bmatrix}
= \begin{bmatrix}
-13 \\
2 \\
7
\end{bmatrix}$


## contoh soal

$\vec{i} = (1, 0, 0), \quad \vec{j} = (0, 1, 0)$

$\vec{i} \times \vec{j} = (0, 0, 1) = \vec{k}$

$- (\vec{j} \times \vec{i}) = (0, 0, -1) = -\vec{k}$
